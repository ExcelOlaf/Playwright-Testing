```python
import os
from dotenv import load_dotenv
from openai import AzureOpenAI
load_dotenv()
```

```python
AZURE_OPENAI_ENDPOINT = "https://dev.devportal.aiapim.jci.com" # Or "https://aiapim.jci.com" for production
AZURE_OPENAI_API_KEY = "nullkey" # This does not matter, since we are using APIM subscription key
AZURE_APIM_API_KEY = "98f05ca74b954c6984434badba550ae2" # This is your subscription key
AZURE_OPENAI_API_DEPLOYMENT = "o3-2025-04-16" # This is the deployment ID of your model 
AZURE_OPENAI_API_VERSION = "2024-02-01"
```

```python
client = AzureOpenAI(
  azure_endpoint = AZURE_OPENAI_ENDPOINT, 
  api_key = AZURE_OPENAI_API_KEY,  
  api_version = AZURE_OPENAI_API_VERSION,
  default_headers = {"Ocp-Apim-Subscription-Key": AZURE_APIM_API_KEY}
)
```

```python
response = client.chat.completions.create(
    model = AZURE_OPENAI_API_DEPLOYMENT,
    messages = [
        {"role": "system", "content": "You are a helpful assistant who talks like a pirate."},
        {"role": "user", "content": "What does executive mean"},
    ]
)

print(response.choices[0].message.content)
```
